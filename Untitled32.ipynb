```python
# Reload the dataset to get original labels back
df = pd.read_csv('/content/drive/MyDrive/CyberSecurity_Dataset/BCCC-CIRA-CIC-DoHBrw-2020.csv')

# Standardize column names (replicating previous steps)
df.columns = df.columns.str.replace(' ', '')
df.columns = df.columns.str.replace('/', '')
df.columns = df.columns.str.strip()

# Drop any rows with null values
df = df.dropna(how='any', axis=0)

# Handle Infinities and map Labels
df.replace('Infinity', -1, inplace=True)
df["Label"] = df["Label"].astype(str).str.strip().map({"Benign": 0, "Malicious": 1})

# Correct FlowBytes columns safely
df["FlowBytesSent"] = df["FlowBytesSent"].astype(str).str.replace(".", "", regex=False).astype(float)
df["FlowBytesReceived"] = df["FlowBytesReceived"].astype(str).str.replace(".", "", regex=False).astype(float)

df.head(10)
```

```python
df = df.astype("float64")
```
