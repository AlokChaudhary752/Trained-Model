```python
import os
import glob
import pandas as pd

# Ensure df is loaded in the current runtime session
if 'df' not in locals() and 'df' not in globals():
    folder_path = '/content/drive/MyDrive/CyberSecurity_Dataset'
    all_files = glob.glob(os.path.join(folder_path, '*.csv'))
    df_list = [pd.read_csv(file, low_memory=False) for file in all_files]
    if df_list:
        df = pd.concat(df_list, axis=0, ignore_index=True)
    else:
        raise FileNotFoundError("No dataset CSV files found in Drive!")

print(df.head())
```

```python
df['Label'] = df['Label'].map({'Benign': 0, 'Malicious': 1})
```

```python
from sklearn.preprocessing import LabelEncoder

le_target = LabelEncoder()
# Converting to string first ensures fit_transform works reliably even if already mapped to numeric values
df['Label'] = le_target.fit_transform(df['Label'].astype(str))
```

```python
print(df.dtypes)
```

```python
num_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
if 'Label' in num_cols:
    num_cols.remove('Label')

print("Columns to be scaled:", num_cols)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])

print(df[num_cols].head())
```

```python
from sklearn.preprocessing import MinMaxScaler

min_max = MinMaxScaler()
df[num_cols] = min_max.fit_transform(df[num_cols])

print(df[num_cols].head())

```
